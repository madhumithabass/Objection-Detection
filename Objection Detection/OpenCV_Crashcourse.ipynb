## 10 Step Framework for OpenCV Object Detection

- Step 1: Import Libraries
- Step 2: Import Model
- Step 3: Import Image/Video
- Step 4: Convert this into GreyScale
- Step 5: Use `MultiScaleDetector` to get the position coordinates.
- Step 6: Use a looping statement to draw rectangular bounding boxes around the objects detected.
- Step 7: Display the Image
- Step 8: Use `waitKey()` to pause the program for a given number of seconds, refresh continuously, and get the key to close.
- Step 9: Use `destroyAllWindows()` to close the video/image frame.
- Step 10: Close all the windows.

```python
import cv2
print(cv2.__version__)
```

```python
front_face_detector = cv2.CascadeClassifier("haarcascade_frontalface_default.xml")

image = cv2.imread("Multiple_images.jpg")
grey_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
face_detector = front_face_detector.detectMultiScale(grey_image)
print(face_detector)

for x,y,w,h in face_detector:
    cv2.rectangle(grey_image,(x,y),(x+w,y+h),(0,0,0),)
cv2.imshow("image_window", grey_image)
cv2.waitKey()
    
```

# CCTV/WebCam Detection

```python
import cv2

front_face_detector = cv2.CascadeClassifier("haarcascade_frontalface_default.xml")

video = cv2.VideoCapture(0)

while True:
    sucess, frames = video.read()
    if sucess == True:
        grey_image = cv2.cvtColor(frames, cv2.COLOR_BGR2GRAY)
        face_detector = front_face_detector.detectMultiScale(grey_image)
        print(face_detector)
    
        for x,y,w,h in face_detector:
            cv2.rectangle(grey_image,(x,y),(x+w,y+h),(0,0,0),)
        
        cv2.imshow("image_window", grey_image)
        key = cv2.waitKey(1)
        if key == 81 or key == 113:
            break

video.release()
cv2.destroyAllWindows()
```
